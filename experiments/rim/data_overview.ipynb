# RIM Experiment Overview

Lists all recorded runs and their key parameters.

```python
import pandas as pd
from rim_teleop.data import ExperimentRegistry

DATA_DIR = "/home/athena/csirois/data/franka/rim"

registry = ExperimentRegistry(DATA_DIR)
runs = registry.list_runs()
print(f"{len(runs)} runs found in {DATA_DIR}")
```

```python
summary_df = pd.DataFrame([r.summary() for r in runs])
summary_df
```

## Filter runs

```python
# Example: only RIM-enabled runs with force feedback
filtered = summary_df[summary_df["rim_enabled"] == True]
filtered
```
